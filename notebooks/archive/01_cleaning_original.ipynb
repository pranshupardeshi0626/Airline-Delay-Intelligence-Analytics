```python
# converting all the required column into proper formate datetime format for futher anaylsis
# in time column there is hour and minutes both so converted them in to_date time column
# in integer column there are only minutes are given so converted them into intergers

# -------- Time columns --------
time_cols = [
    'crs_dep_time',
    'dep_time',
    'wheels_off',
    'wheels_on',
    'crs_arr_time',
    'arr_time'
]

# Convert float -> Int64 -> string -> pad zeros
df[time_cols] = (
    df[time_cols]
    .astype('Int64')
    .astype(str)
    .apply(lambda x: x.str.zfill(4))
)

# Convert to datetime
for col in time_cols:
    df[col] = pd.to_datetime(
        df[col],
        format='%H%M',
        errors='coerce'
    )


# -------- Integer columns --------
int_cols = [
    'taxi_out',
    'taxi_in',
    'arr_delay',
    'distance',
    'crs_elapsed_time',
    'actual_elapsed_time',
]

df[int_cols] = df[int_cols].astype('Int64')
```

```python
# flight date from int into datetime format
df['fl_date'] = pd.to_datetime(df['fl_date'], errors='coerce')
```

```python
# carrier flight number into string
df['op_carrier_fl_num'] = df['op_carrier_fl_num'].astype(str)
```

```python
# departure delay column from float into int
df['dep_delay'] = df['dep_delay'].astype("Int64")
```

```python
# Actual early arrival delay values
df['arrived_early'] = np.where((df['arr_delay'].notna()) & (df['arr_delay'] < 0),df['arr_delay'],np.nan)

# Actual late arrival delay values
df['arrived_late'] = np.where((df['arr_delay'].notna()) & (df['arr_delay'] > 0),df['arr_delay'],np.nan)
```

```python
# Create arrival status column
df['arrival_status'] = 'Unknown'

# Cancelled flights
df.loc[df['cancelled'] == 1, 'arrival_status'] = 'Cancelled'

# Diverted flights
df.loc[df['diverted'] == 1, 'arrival_status'] = 'Diverted'

# Normal flights classification
df.loc[(df['cancelled']==0) & (df['diverted']==0) & (df['arr_delay'] < 0),'arrival_status'] = 'Arrived Early'

df.loc[(df['cancelled']==0) & (df['diverted']==0) & (df['arr_delay'] == 0),'arrival_status'] = 'Arrived On Time'

df.loc[(df['cancelled']==0) & (df['diverted']==0) & (df['arr_delay'] > 0),'arrival_status'] = 'Arrived Late'
```

```python
df['arrival_status'].unique()
```

```python
# Actual early departure delay values
df['departed_early'] = np.where((df['dep_delay'].notna()) & (df['dep_delay'] < 0),df['dep_delay'],np.nan)

# Actual late departure delay values
df['departed_late'] = np.where((df['dep_delay'].notna()) & (df['dep_delay'] > 0),df['dep_delay'],np.nan)
```

```python
# Create departure status column
df['departure_status'] = 'Unknown'

# Cancelled flights
df.loc[df['cancelled'] == 1, 'departure_status'] = 'Cancelled'

# Diverted flights
df.loc[df['diverted'] == 1, 'departure_status'] = 'Diverted'

# Normal flights classification
df.loc[(df['cancelled']==0) & (df['diverted']==0) & (df['dep_delay'] < 0),'departure_status'] = 'departed Early'

df.loc[(df['cancelled']==0) & (df['diverted']==0) & (df['dep_delay'] == 0),'departure_status'] = 'departed On Time'

df.loc[(df['cancelled']==0) & (df['diverted']==0) & (df['dep_delay'] > 0),'departure_status'] = 'departed Late'
```

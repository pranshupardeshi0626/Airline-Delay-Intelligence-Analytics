```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import numpy as np
```

#  Univariate Analysis

```python
df = pd.read_csv("flight_cleaned_for_visualization.csv",low_memory=False)
```

```python
df.info()
```

```python
airline_map = {
    'AA': 'American Airlines',
    'AS': 'Alaska Airlines',
    'B6': 'JetBlue Airways',
    'DL': 'Delta Air Lines',
    'F9': 'Frontier Airlines',
    'G4': 'Allegiant Air',       
    'HA': 'Hawaiian Airlines',
    'MQ': 'Envoy Air',
    'NK': 'Spirit Airlines',
    'OH': 'PSA Airlines',
    'OO': 'SkyWest Airlines',
    'UA': 'United Airlines',
    'WN': 'Southwest Airlines',
    'YX': 'Republic Airways',
    '9E': 'Endeavor Air'
}
```

```python
df['airline_full'] = df['op_unique_carrier'].map(airline_map)
```

```python
print(df[df['airline_full'].isnull()]['op_unique_carrier'].unique())
```

```python
# checked for flight distribution
fig = px.treemap(
    df,
    path=['airline_full'],
    title='Flight Volume Share by Airline',
    color='airline_full',
    color_discrete_sequence=px.colors.qualitative.Bold
)

fig.update_traces(
    textinfo='label+percent entry'
)

fig.update_layout(
    width=1000,
    height=650
)

fig.show()
```

```python
# month dirstribution
month_counts = df['month'].value_counts().sort_index()

fig = px.pie(
    values=month_counts.values,
    names=month_counts.index,
    hole=0.45,   # makes it donut
    title='Flight Distribution Across Months',
    color_discrete_sequence=px.colors.qualitative.Set3
)

fig.update_traces(
    textinfo='label+percent'
)

fig.update_layout(
    width=850,
    height=650
)

fig.show()
```

Flight records are uniformly distributed across all 12 months.

This indicates the dataset is seasonally balanced and not biased toward a specific travel period.

Therefore further delay and cancellation analysis across months can be considered reliable.

```python
# arrival distribution
arrival_counts = df['arrival_status'].value_counts()

fig = px.funnel(
    x=arrival_counts.values,
    y=arrival_counts.index,
    title='Arrival Status Distribution',
    color=arrival_counts.index,
    color_discrete_sequence=px.colors.qualitative.Pastel
)

fig.update_layout(
    width=900,
    height=600
)

fig.show()
```

```python
# checked for departure status distribution
departure_counts = df['departure_status'].value_counts()

fig = px.bar(
    x = departure_counts.index,
    y = departure_counts.values,
    text = departure_counts.values,
    color = departure_counts.index,
    color_discrete_sequence = px.colors.qualitative.Dark24,
    labels={'x':'Flight departure Status', 'y':'Number of Flights'},
    title = 'Departure Status Distribution'
)
fig.update_traces(
    textposition = 'outside'
)

fig.update_layout(
    bargap = 0.12,
    width = 900,
    height = 550
)

fig.update_yaxes(
    range = [0, max(departure_counts.values)*1.15]
)
fig.show()
    
```

```python
# created weekend map
weekend_map = {
    0:'Weekday',
    1:'Weekend'
}

df['day_type'] = df['is_weekend'].map(weekend_map)
```

```python
# weekend vs weekday flight distribution
weekend_counts = df['day_type'].value_counts()

fig = px.pie(
    values=weekend_counts.values,
    names=weekend_counts.index,
    hole=0.5,
    title='Weekday vs Weekend Flight Distribution',
    color_discrete_sequence=['royalblue','orange']
)

fig.update_traces(
    textinfo='label+percent'
)

fig.update_layout(
    width=800,
    height=650
)

fig.show()
```

```python
# checked for distribution of week
day_counts = df['day_of_week'].value_counts().sort_index()

fig = px.bar(
    x = day_counts.index,
    y = day_counts.values,
    text = day_counts.values,
    color = day_counts.index,
    color_discrete_sequence = px.colors.qualitative.Set3,
    title = 'Flight Distribution by Day of Week',
    labels = {'x' : 'Day of Week', 'y':'Number of Flights'}
)
fig.update_traces(
    textposition = 'outside'
)

fig.update_layout(
    bargap = 0.12,
    width = 600,
    height = 550
)

fig.update_yaxes(
    range = [0, max(day_counts.values)*1.15]
)
fig.show()
    
```

## Multivariate Analysis

```python
delay_by_airline = df.groupby('op_unique_carrier')[
    ['weather_delay','carrier_delay','nas_delay','late_aircraft_delay']
].mean().reset_index()

fig = px.bar(
    delay_by_airline,
    x='op_unique_carrier',
    y=['weather_delay','carrier_delay','nas_delay','late_aircraft_delay'],
    title='Average Delay Type Contribution by Airline',
    labels={
        'value':'Average Delay (Minutes)',
        'op_unique_carrier':'Airline',
        'variable':'Delay Type'
    },
    color_discrete_sequence=px.colors.qualitative.Bold
)

fig.update_layout(
    width=1000,
    height=700
)

fig.show()
```

```python
# departure hour,month,average departure delay
import plotly.express as px

heatmap_data = df.pivot_table(
    values='dep_delay',
    index='month',
    columns='scheduled_dep_hour',
    aggfunc='mean'
)

fig = px.imshow(
    heatmap_data,
    text_auto='.1f',
    aspect='auto',
    color_continuous_scale='RdYlBu_r',
    title='Average Departure Delay by Month and Scheduled Departure Hour',
    labels={
        'x':'Scheduled Departure Hour',
        'y':'Month',
        'color':'Avg Delay'
    }
)

fig.update_layout(
    width=1000,
    height=700
)

fig.show()
```

```python
# Airline,Month,Cancellation rate
cancel_airline_month = df.pivot_table(
    values='cancelled',
    index='op_unique_carrier',
    columns='month',
    aggfunc='mean'
) * 100

fig = px.imshow(
    cancel_airline_month,
    text_auto='.1f',
    aspect='auto',
    color_continuous_scale='Inferno',
    title='Cancellation Rate by Airline and Month',
    labels={
        'x':'Month',
        'y':'Airline',
        'color':'Cancellation %'
    }
)

fig.update_layout(
    width=1000,
    height=700,
    title_x=0.5
)

fig.show()
```

```python
corr_cols = [
    'dep_delay',
    'arr_delay',
    'weather_delay',
    'carrier_delay',
    'nas_delay',
    'late_aircraft_delay',
    'taxi_out',
    'taxi_in',
    'air_time',
    'distance'
]

corr_matrix = df[corr_cols].corr()

fig = px.imshow(
    corr_matrix,
    text_auto='.2f',
    aspect='auto',
    color_continuous_scale='RdBu_r',
    title='Correlation Heatmap of Flight Delay Factors'
)

fig.update_layout(
    width=1000,
    height=800,
    title_x=0.5
)

fig.show()
```

```python
# checked for month+ delay type + average delay contribution
delay_month = df.groupby('month')[
    ['weather_delay','carrier_delay','nas_delay','late_aircraft_delay']
].mean().reset_index()

fig = px.bar(
    delay_month,
    x='month',
    y=['weather_delay','carrier_delay','nas_delay','late_aircraft_delay'],
    title='Average Delay Type Contribution Across Months',
    labels={
        'month':'Month',
        'value':'Average Delay (Minutes)',
        'variable':'Delay Type'
    },
    color_discrete_sequence=px.colors.qualitative.Set1
)

fig.update_layout(
    width=1000,
    height=700,
    title_x=0.5
)

fig.show()
```
